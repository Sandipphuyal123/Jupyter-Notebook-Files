### Creating a DataFrame

```python
# i am creating a dataset about the cities in nepal and their information
city=['Damak', 'pokhara', 'biratchowk', 'jirikhemti','Taulihawa','kathmandu']
population=[107227, 513504, 120213, 2682, 8887,862400]
area=[75.74,464.2,None,100,136.91,49.45]
malePopulation_per=[47.4,48.2,47.4,48.8,49.5,50.8]
malePopulation=[50819, 247495, 57018, 1288,43998,424144]
femalePopulation=[56408, 266009, 63195, 1394,44876, 424144]
femalePopulation_per=[52.6,51.8,52.6,52.2,50.5,49.2]
district=['Jhapa','Kaski','Jhapa','Terhathum','Kapilvastu','Kathmandu']
province=['Koshi', 'Gandaki', 'Koshi', 'Koshi','Lumbini','Bagmati']
mayor=['Ram Kumar', 'Dhana Raj Acharya', 'Sundar Haraicha', 'sanjay kumar','Sudeep Paudel','Balendra Shah']
osm_id=['7405325', '6104486', '34830836', '454477382','14978816','14978816']
```

```python
# this is the dictionary that makes the list into a dataframe.
dict_city={'City':city, 
           'Population':population,
           'Area': area, 
           'Male_Population_Percentage':malePopulation_per,
           'Male_Population':malePopulation,
           'Female_Population_Percentage':femalePopulation_per,
           'Female_Population':femalePopulation,
           'District':district,
           'Province': province,
           'Mayor_2025':mayor,
           'OSM_ID':osm_id
}
```

```python
df_city=pd.DataFrame(dict_city) # Linking the dataFrame to a variable for easier access
df_city
```

### Selecting Two or more Columns 

```python
df_city
```

```python
df_city.isnull().sum()
```

```python
df_city['Gender_Total_%']=df_city['Male_Population_Percentage']+df_city['Female_Population_Percentage']
df_city
```

```python
df_city['Population_Density']=(df_city['Population']/df_city['Area']).round(2)
df_city
```

```python
df_city['Male_population']=(df_city['Population']*df_city['Male_Population_Percentage']/100).round(2)
df_city['Female_population']=(df_city['Population']*df_city['Female_Population_Percentage']/100).round(2)
df_city
```

```python
# MOST POPULATED CITY

df_city.loc[df_city['Population'].idxmax(), ['City', 'Population']]

```

```python
# MOST DENSE CITY

df_city.loc[df_city['Population_Density'].idxmax(), ['City','Population_Density']]
```

```python
plt.figure(figsize=(10,5))
plt.bar(df_city['City'], df_city['Population'])
plt.title('Population by CITY')
plt.xlabel('City')
plt.ylabel('Population')
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.bar(df_city['City'], df_city['Population_Density'])
plt.title('Population Density by City')
plt.xlabel('City')
plt.ylabel('People per sq.km')
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.bar(df_city['City'], df_city['Male_Population'], label='Male')
plt.bar(df_city['City'], df_city['Female_Population'],
        bottom=df_city['Male_Population'], label='Female')

plt.title('Gender Distribution by City')
plt.xlabel('City')
plt.ylabel('Population')
plt.legend()
plt.show()
```
