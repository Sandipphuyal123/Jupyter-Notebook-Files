### Creating a DataFrame

```python
# i am creating a dataset about the cities in nepal and their information
city=['Damak', 'pokhara', 'biratchowk', 'jirikhemti']
population=[107227, 513504, 120213, 2682]
area=[75.74,464.2,None,100]
malePopulation_per=[47.4,48.2,47.4,48.8]
femalePopulation_per=[52.6,51.8,52.6,52.2]
district=['Jhapa','Kaski','Jhapa','Terhathum']
province=['Koshi', 'Gandaki', 'Koshi', 'Koshi']
mayor=['Ram Kumar', 'Dhana Raj Acharya', 'Sundar Haraicha', 'sanjay kumar']
```

```python
# this is the directory that makes the list into a dataframe.
dict_city={'City':city, 
           'Population':population,
           'Area': area, 
           'Male_Population_Percentage':malePopulation_per, 
           'Female_Population_Percentage':femalePopulation_per,
           'District':district,
           'Province': province,
           'Mayor_2025':mayor
}
```

```python
df_city=pd.DataFrame(dict_city) # Linking the dataFrame to a variable for easier access
df_city
```
