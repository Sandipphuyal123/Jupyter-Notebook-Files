```python
# I want to narrow down the columns so i can focus on the dataFrame.
df_subset=df[['country','day','time','cases_new','cases_1M_pop','deaths_1M_pop','population','cases_critical']]
df_subset
```

```python
df.head()
```

```python
#  Redudant columns, i removed 'time' because the data in this dataframe might not be consistent cuz 
#  COVID-19 data changes from days and weeks not hours[! the insights from the 'time[hours]' might not be reliable] 
df=df.drop(columns=['Unnamed: 0', 'index','time'])
df.head()
```

```python
# making 'day' column from text->datetime data and deaths_new from text to numeric
df['day']=pd.to_datetime(df['day'], errors='coerce')
df['deaths_new']=pd.to_numeric(df['deaths_new'], errors='coerce')
```

```python
# sorting 'day' into ascending order.
df=df.sort_values(by=['country', 'day'])
df.head()
```

```python
# we are trying to group[DataFrame] the countries to subtable, so the data analysis doesn't go out of the
#  environment. Since populaiton of the country doesn't change day to day. we can make the population stagnent.

df['population']=df.groupby('country')['population'].ffill().bfill()
df
```

# Creating new series for further explanation of the dataset
- for 


```python
# we need case fatality_rate to get insights between countries. 
# we will see NaN in the fatality rate because in the earlier days of COVID there were small entries with
# - case_total and deaths_total.
df['fatality_rate']=df['deaths_total']/df['cases_total']
df
```

```python
df['death_per_active_case']=df['deaths_new']/ df['cases_active']
```

```python
df.head()
```

```python
df['deaths_per_million_calc']=(df['deaths_total']/df['populaiton'])*1000000
df.head()
```

```python
# we are making an average of 7 days case because there is information inconsistency with the cases
# due to various factor like weekends and information delay.
df['Avg_7day_cases']=(df.groupby('country')['cases_new']
    .rolling(7).mean().reset_index(level=0, drop=True))
df.head(2)
```
