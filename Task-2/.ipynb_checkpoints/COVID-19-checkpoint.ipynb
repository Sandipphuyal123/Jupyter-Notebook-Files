# Data Cleaning

### Problems Found in the DataSet
- These column are not needed ( Unnamed: 0, index, time )
- Changing ['date'] column from 'text' to 'datetime' dtype.
- Changing ['deaths_new'] column from 'text' to 'numeric'
- There are some NaN in fixed columns ( country, continent, population ).
- A column Shift Error has occured in continent column causing some continents name to be population and populaiton to be NaN.[i.e 876, 1789,3395,5032]
- There are 229 countries in the dataset, since the official number is 195. We can drop the rest.
- In the continent, Austrilia continent is switched to Oceania since misunderstanding can be made because of Austrilia state and Austrilia continent in matplot.
- Removed all the same days Data and left one row for one day, Late time data of the day for accurate data
- 

```python
# -Datatypes: we have day and time as text not time-series data[error]. 
#  we have deaths_new as a obj[text], make it numeric so u can do calculations on that columns.
df.info()
```

```python
# looks like Unnamed, index and time is redudant columns, i can remove them later...
df.head()
```

```python
# I want to narrow down the columns so i can focus on the dataFrame.
df_subset=df[['country','day','cases_new','cases_1M_pop','deaths_1M_pop','population','cases_critical']]
df_subset.head()
```

```python
#  Redudant columns, i removed 'time' because the data in this dataframe might not be consistent cuz 
#  COVID-19 data changes from days and weeks not hours[! the insights from the 'time[hours]' might not be reliable] 
df=df.drop(columns=['Unnamed: 0', 'index','time'])
```

```python
# making 'day' column from text->datetime data and deaths_new from text to numeric
df['day']=pd.to_datetime(df['day'], errors='coerce')
df['deaths_new']=pd.to_numeric(df['deaths_new'], errors='coerce')
```

```python
#continent is a constant value it shouldn't have nullvalues, can use ffill and bfill. 
df.isnull().sum()
```

```python
# Error found continent, missing Austrilia.
df['continent'].value_counts()
```

```python
# Error found [Column Shift Error] which caused some population to be NaN and continent to be population.
# looks like we need to make Australia/Oceania into 'Oceania' only so that it doesn't conflicit.
df.groupby('continent')['country'].unique()
```

```python
row_iloc = df.iloc[876]

print("\nRow selected by .iloc[1]:")
print(row_iloc)
```

```python
# Finding out which continents are corrupted.
# we can find this by making the continent numeric, since the column[continent]- 
# -where error is in numeric we can ignore the string ones.
df.shape
```

```python
# lets first find the columns before changing the dataset.
num_continent=pd.to_numeric(df['continent'],errors='coerce')
bad_data_mask = num_continent.notnull()
df_corruptedContinents=df[bad_data_mask]
df_corruptedContinents.head()
```

```python
# solving the Column Shift Error
df = df[num_continent.isna()].copy()
df
```

```python
# Changing the name of Austrilia continent to Oceania since austrilia name is both the state and the continent.
df.loc[df['continent']=='Austrilia/Oceania','continent']='Oceania'
```

```python
# mapping
real_continents=['Asia','Europe','North-America','South-America','Africa','Oceania']
continent_map = (
    df[df['continent'].isin(real_continents)]
    .drop_duplicates('country')
    .set_index('country')['continent']
    .to_dict()
)
```

```python
df['continent'] = df['country'].map(continent_map)
```

```python
# Grouping the countries so the calculation of countries doesn't go outside of its individual country.
# filling up populaiton with ffill and bfill
df['population']=df.groupby('country')['population'].ffill().bfill()
df.head(3)
```

```python
# This drops rows that don't have a valid continent/population (like 'World' or errors)
df= df.dropna(subset=['continent', 'population'])
```

```python
print(f"Dataset cleaned. New shape: {df.shape}")
print(df['continent'].value_counts())
```

```python
global_daily = (
    df.groupby('day')[['cases_total', 'deaths_total']]
    .sum()
    .reset_index()
)
```

```python
peak_cases_row = global_daily.loc[global_daily['cases_total'].idxmax()]
peak_deaths_row = global_daily.loc[global_daily['deaths_total'].idxmax()]
```

```python
plt.figure(figsize=(12,6))

# Main lines
plt.plot(
    global_daily['day'],
    global_daily['cases_total'],
    label=f"Total Cases (Peak: {int(peak_cases_row['cases_total']):,})",
    linewidth=2
)

plt.plot(
    global_daily['day'],
    global_daily['deaths_total'],
    label=f"Total Deaths (Peak: {int(peak_deaths_row['deaths_total']):,})",
    linewidth=2
)

# Mark peak points
plt.scatter(
    peak_cases_row['day'],
    peak_cases_row['cases_total'],
    s=90
)

plt.scatter(
    peak_deaths_row['day'],
    peak_deaths_row['deaths_total'],
    s=90
)

# Annotate Peak Total Cases
plt.annotate(
    f"PEAK TOTAL CASES\n{int(peak_cases_row['cases_total']):,}",
    xy=(peak_cases_row['day'], peak_cases_row['cases_total']),
    xytext=(peak_cases_row['day'], peak_cases_row['cases_total'] * 0.75),
    arrowprops=dict(arrowstyle='->', linewidth=1.5),
    fontsize=11,
    fontweight='bold',
    color='darkblue'
)

# Annotate Peak Total Deaths (angled & offset)
plt.annotate(
    f"PEAK TOTAL DEATHS\n{int(peak_deaths_row['deaths_total']):,}",
    xy=(peak_deaths_row['day'], peak_deaths_row['deaths_total']),
    xytext=(peak_deaths_row['day'] - pd.Timedelta(days=120),
            peak_deaths_row['deaths_total'] * 1.5), #2.5
    arrowprops=dict(arrowstyle='->', linewidth=1.5),
    fontsize=11,
    fontweight='bold',
    color='darkred'
)

# Titles and labels
plt.title(
    'Global COVID-19 Total Cases and Deaths Over Time',
    fontsize=14,
    fontweight='bold'
)

plt.xlabel(
    'DATE',
    fontsize=12,
    fontweight='bold'
)

plt.ylabel(
    'COUNT',
    fontsize=12,
    fontweight='bold'
)

# Grid & legend
plt.legend(fontsize=10)
plt.grid(True)

plt.show()
```

```python
# 1. Create a year column from the 'day' column
df['year'] = df['day'].dt.year

# 2. Filter for the main three years (2020, 2021, 2022)
df_waves = df[df['year'].isin([2020, 2021, 2022])].copy()

# 3. Calculate total deaths per year
# Note: Using sum of 'deaths_new' to get total annual deaths
yearly_deaths = df_waves.groupby('year')['deaths_new'].sum().reset_index()

# 4. Plotting the data
plt.figure(figsize=(10, 6))
sns.set_style("whitegrid")

# Create the bar chart
bars = plt.bar(
    yearly_deaths['year'].astype(str), 
    yearly_deaths['deaths_new'], 
    color=['#3498db', '#e74c3c', '#2ecc71'], # Distinct colors for each year
    edgecolor='black'
)

# Adding labels and title
plt.title('Comparison of Total COVID-19 Deaths by Year (2020–2022)', fontsize=14, fontweight='bold')
plt.xlabel('YEAR', fontsize=12, fontweight='bold')
plt.ylabel('TOTAL DEATHS (Millions)', fontsize=12, fontweight='bold')

# Formatting the y-axis to be more readable
plt.ticklabel_format(style='plain', axis='y')

# Add data labels on top of the bars for clarity
for bar in bars:
    yval = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2, 
        yval + (yval * 0.01), 
        f'{int(yval):,}', 
        ha='center', 
        va='bottom', 
        fontweight='bold',
        fontsize=11
    )

plt.tight_layout()
plt.show()

# Print the numeric summary for verification
for index, row in yearly_deaths.iterrows():
    print(f"Total deaths in {int(row['year'])}: {int(row['deaths_new']):,}")
```

# Formula:
    CFR (%) = ( Total Deaths / Total Cases ) × 100

| Country    | Likely Causes for High CFR                                                                                                                |
| ---------- | ----------------------------------------------------------------------------------------------------------------------------------------- |
| **Yemen**  | Severe healthcare collapse due to **ongoing conflict**, very limited hospital access, low testing (underreporting of cases → higher CFR). |
| **Sudan**  | Political instability, weak health infrastructure, poor access to COVID care.                                                             |
| **Peru**   | Initial mismanagement + overwhelmed hospitals, high comorbidity prevalence (diabetes, obesity).                                           |
| **Mexico** | High prevalence of chronic conditions, hospital overload during Delta wave, under-testing inflates CFR.                                   |
| **Syria**  | Civil war + conflict → collapsed healthcare, limited testing, underreported infections.                                                   |


```python
df['fatality_rate'] = (
    df['deaths_total'] / latest_df['cases_total']
) * 100

# Remove invalid values
df = df.replace([np.inf, -np.inf], np.nan)
df = df.dropna(subset=['fatality_rate'])
```

```python
# --- Top 5 Countries by Fatality Rate ---
top_fatality_5 = (
    df
    .sort_values('fatality_rate', ascending=False)
    .head(5)
)

plt.figure(figsize=(10,6))
sns.set_style("whitegrid")

bars = plt.bar(
    top_fatality_5['country'],
    top_fatality_5['fatality_rate'],
    color='#8e44ad',  # Purple
    edgecolor='black'
)

# Titles and labels
plt.title(
    'Top 5 Countries by COVID-19 Case Fatality Rate',
    fontsize=14,
    fontweight='bold'
)
plt.xlabel('COUNTRY', fontsize=12, fontweight='bold')
plt.ylabel('CASE FATALITY RATE (%)', fontsize=12, fontweight='bold')

# Annotate the bars with percentages
for bar in bars:
    y = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        y + y*0.03,
        f'{y:.2f}%',
        ha='center',
        va='bottom',
        fontweight='bold',
        fontsize=11
    )

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn.linear_model import LinearRegression

# Continent-level metrics (latest date per continent)
continent_metrics = df.groupby('continent').agg({
    'deaths_total':'max',
    'cases_total':'max',
    'population':'sum',
    'tests_total':'max'  # assuming 'tests_total' exists
}).reset_index()

# Calculate fatality rate and tests per million
continent_metrics['fatality_rate'] = (continent_metrics['deaths_total'] / continent_metrics['cases_total']) * 100
continent_metrics['tests_1M_pop'] = (continent_metrics['tests_total'] / continent_metrics['population']) * 1e6

# --- Bubble Plot ---
plt.figure(figsize=(12,7))
sns.set_style("whitegrid")

# Scatter bubbles
plt.scatter(
    continent_metrics['tests_1M_pop'],
    continent_metrics['fatality_rate'],
    s=np.sqrt(continent_metrics['population'])/200,
    c=range(len(continent_metrics)),
    cmap='tab10',
    alpha=0.8,
    edgecolors='black'
)

# Annotate each bubble
for i, row in continent_metrics.iterrows():
    plt.text(
        row['tests_1M_pop'] + 5000,  # small offset
        row['fatality_rate'],
        f"{row['continent']}\nPop:{int(row['population']/1e6)}M",
        fontsize=10,
        fontweight='bold'
    )

# --- Regression line (trend) ---
X = continent_metrics['tests_1M_pop'].values.reshape(-1,1)
y = continent_metrics['fatality_rate'].values
reg = LinearRegression().fit(X, y)
x_range = np.linspace(X.min()*0.9, X.max()*1.1, 100)
y_pred = reg.predict(x_range.reshape(-1,1))
plt.plot(x_range, y_pred, color='red', linestyle='--', linewidth=2, label='Trend: Testing vs Fatality')

# Titles & labels
plt.title('COVID-19: Fatality Rate vs Testing by Continent', fontsize=14, fontweight='bold')
plt.xlabel('Tests per 1M Population', fontsize=12, fontweight='bold')
plt.ylabel('Case Fatality Rate (%)', fontsize=12, fontweight='bold')
plt.legend()
plt.grid(True, linestyle='--', linewidth=0.5)
plt.tight_layout()
plt.show()
```
